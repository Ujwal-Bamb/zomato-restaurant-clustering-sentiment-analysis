### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1

Is there a significant relationship between a restaurant's average meal cost and its average customer rating?

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**H₀:** There is no significant relationship between restaurant's average meal cost and its average customer rating ** (ρ = 0)**

**H₁:** There is a significant relationship between restaurant's average meal cost and its average customer rating **(ρ ≠ 0).**

**Type of Test :** Two-tailed Spearman Rank Correlation test.

**Significance level (α):** 0.05

````python
# Perform Statistical Test to obtain P-Value
restaurant_rating = (review.groupby('Restaurant')['Rating'].mean().reset_index())
rating_cost = restaurant_rating.merge(restaurant[['Name', 'Cost']],left_on='Restaurant',right_on='Name',how='inner')


from scipy.stats import spearmanr
correlation, p_value = spearmanr(rating_cost['Cost'], rating_cost['Rating'])
print(f"Correlation: {correlation:.4f}")
print(f"P-value: {p_value:.4f}")

alpha = 0.05

if p_value < alpha:
    print("Reject H₀")
    print("There is a statistically significant relationship between restaurant cost and average customer rating.")
else:
    print("Fail to Reject H₀")
    print("There is no statistically significant relationship between restaurant cost and average customer rating.")
````

##### Which statistical test have you done to obtain P-Value?

I performed the **Spearman Rank Correlation test** to determine whether there is a statistically significant relationship between restaurant  average meal cost and its average customer rating

The test resulted in a **Spearman correlation coefficient of 0.3826** and a **p-value of 0.0001.** Since the p-value is less than 0.05, we **reject the null hypothesis** and conclude that there is a statistically significant relationship between restaurant cost and average customer rating.

This was a two-tailed test because the alternative hypothesis states that a relationship exists without specifying whether it should be positive or negative.

##### Why did you choose the specific statistical test?

````python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=rating_cost,x='Cost',y='Rating')

plt.xlabel('Restaurant Average Meal Cost')
plt.ylabel('Average Restaurant Rating')
plt.title('Relationship Between Restaurant Cost and Average Restaurant Rating')

plt.show()
````

I chose the Spearman Rank Correlation test because the objective was to determine whether restaurant cost and average restaurant rating are related.

Before performing the statistical test, I used a scatter plot to visually examine the relationship between cost and average rating. The scatter plot did not show a clear linear pattern, so instead of assuming a linear relationship, Spearman correlation is suitable because it measures the strength and direction of association without requiring the variables to follow a normal distribution.

Therefore, Spearman correlation was appropriate for testing the relationship between restaurant cost and average restaurant rating.

### Hypothetical Statement - 2

Is the proportion of highly rated reviews (rating ≥ 4) significantly greater than 50%?

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**H₀ (Null Hypothesis):**

The proportion of highly rated reviews is 50%.

$$
H_0: p = 0.50
$$

**H₁ (Alternative Hypothesis):**

The proportion of highly rated reviews is **greater than 50%**.

$$
H_1: p > 0.50
$$.

**Type of Test :** One-sample proportion z-test (right-tailed)

#### 2. Perform an appropriate statistical test.

````python
high_rated = (review['Rating'] >= 4).sum()
total_rated = review['Rating'].notna().sum()

proportion = high_rated / total_rated

print("Highly Rated Reviews:", high_rated)
print("Total Rated Reviews:", total_rated)
print(f"Proportion: {proportion:.4f} ({proportion * 100:.2f}%)")
````

````python
# Perform Statistical Test to obtain P-Value
from statsmodels.stats.proportion import proportions_ztest

count = high_rated
total_reviews = total_rated

z_stat, p_value = proportions_ztest(count,total_reviews,value=0.50,alternative='larger')

print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject H₀")
    print("Conclusion: The proportion of highly rated reviews is significantly greater than 50%.")
else:
    print("Decision: Fail to Reject H₀")
    print("Conclusion: There is not enough evidence to conclude that the proportion of highly rated reviews is greater than 50%.")
````

##### Which statistical test have you done to obtain P-Value?

I performed a One-Sample Proportion Z-Test to determine whether the proportion of highly rated reviews (Rating ≥ 4) is significantly greater than the hypothesized proportion of 50%.


##### Why did you choose the specific statistical test?

I chose the One-Sample Proportion Z-Test because the hypothesis compares the observed proportion of highly rated reviews in the dataset with a specific hypothesized population proportion of 50%.

The reviews were divided into two categories based on the rating: Highly Rated (Rating ≥ 4) and Not Highly Rated (Rating < 4). Since the dataset contains a large number of rated reviews (9,961), the proportion Z-test is appropriate for testing whether the observed proportion of 62.99% is significantly greater than 50%.

````python
# Remove URLs & Remove words and digits contain digits
review['Review_Clean'] = review['Review_Clean'].apply(
    lambda x: re.sub(r'http\S+|www\S+', '', x)
)

# Remove words containing digits and standalone numbers
review['Review_Clean'] = review['Review_Clean'].apply(
    lambda x: re.sub(r'\S*\d\S*', '', x)
)

# Remove extra whitespaces
review['Review_Clean'] = review['Review_Clean'].str.replace(
    r'\s+', ' ', regex=True
).str.strip()

review[['Review', 'Review_Clean']].head(10)
````

We removed URLs and words or tokens containing digits because they are generally less useful for our basic sentiment analysis and may introduce unnecessary noise into the text representation.

#### 5. Removing Stopwords & Removing White spaces

````python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

# Keep important negation words for sentiment analysis
stop_words -= {'no', 'not', 'nor', 'never'}

# Remove Stopwords
review['Review_Clean'] = review['Review_Clean'].apply(
    lambda x: ' '.join(
        word for word in x.split()
        if word not in stop_words
    )
)

# Remove White Spaces
review['Review_Clean'] = review['Review_Clean'].str.replace(
    r'\s+', ' ', regex=True
).str.strip()

review[['Review', 'Review_Clean']].head(10)
````

We removed common English stopwords from the review text to reduce frequently occurring words that provide limited information for sentiment analysis. Extra whitespaces were also removed to maintain a clean and consistent text format.

Instead, we standardized the text through preprocessing steps such as contraction expansion, lowercasing, punctuation removal, removal of unwanted tokens, stopword handling, and whitespace normalization. This preserves the original meaning while making the text suitable for NLP processing.

#### 7. Tokenization

### Clustering Model 1 - K-Means Clustering

````python
# Elbow Method + Silhouette Score


from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

inertia = []
silhouette_scores = []

# Test different numbers of clusters
for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    cluster_labels = kmeans.fit_predict(restaurant_pca)

    # Elbow Method
    inertia.append(kmeans.inertia_)

    # Silhouette Score
    score = silhouette_score(
        restaurant_pca,
        cluster_labels
    )
    silhouette_scores.append(score)


# Create subplots
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# ─ Elbow Method
axes[0].plot(range(2, 11),inertia,marker='o')

axes[0].set_xlabel('Number of Clusters (K)')
axes[0].set_ylabel('Inertia')
axes[0].set_title('Elbow Method')
axes[0].set_xticks(range(2, 11))
axes[0].grid(True, linestyle='--', alpha=0.5)


# ─ Silhouette Score
axes[1].plot(
    range(2, 11),
    silhouette_scores,
    marker='o'
)

axes[1].set_xlabel('Number of Clusters (K)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Score')
axes[1].set_xticks(range(2, 11))
axes[1].grid(True, linestyle='--', alpha=0.5)


plt.suptitle(
    'K-Means Cluster Evaluation',
    fontsize=14
)

plt.tight_layout()
plt.show()


# Print Silhouette Scores
print("Silhouette Scores:")
for k, score in zip(range(2, 11), silhouette_scores):
    print(f"K = {k}: Silhouette Score = {score:.4f}")
````

The Elbow Method did not provide a clearly defined elbow point, so Silhouette Score was used as an additional evaluation measure. Among the tested values from K = 2 to K = 10, K = 9 produced the highest Silhouette Score of 0.1543. Therefore, K = 9 was selected for the K-Means model. The relatively low Silhouette Score indicates that the resulting restaurant segments have some overlap and are not strongly separated.

#### Final K-Means model

````python
from sklearn.cluster import KMeans

# Final K-Means model
kmeans = KMeans(
    n_clusters=9,
    random_state=42,
    n_init=10
)

# Assign clusters to restaurants
restaurant['Cluster'] = kmeans.fit_predict(restaurant_pca)

print("Final Number of Clusters:", restaurant['Cluster'].nunique())

print("\nCluster Distribution:")
print(restaurant['Cluster'].value_counts().sort_index())

from sklearn.metrics import silhouette_score

# Final Silhouette Score
final_silhouette = silhouette_score(
    restaurant_pca,
    restaurant['Cluster']
)

print(f"\nFinal Silhouette Score: {final_silhouette:.4f}")
````

The Elbow Method was used to evaluate different values of **K from 2 to 10**. The Elbow curve did not show a clearly defined elbow point, so the **Silhouette Score** was also considered for selecting the number of clusters.

Among the tested values, **K = 9 achieved the highest Silhouette Score of 0.1543**. Therefore, **9 clusters were selected for the final K-Means model**. The relatively low Silhouette Score indicates that there is some overlap between the clusters, which will be considered while interpreting the cluster results.

#### Create Dendrogram

````python
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt

linkage_matrix = linkage(restaurant_pca, method='ward')

plt.figure(figsize=(16, 6))
dendrogram(linkage_matrix, truncate_mode='lastp', p=20)

plt.title('Dendrogram - Agglomerative Hierarchical Clustering')
plt.xlabel('Restaurants')
plt.ylabel('Distance')
plt.show()
````

### 3. Explain the model which you have used and the feature importance using any model explainability tool?

The final model is **Logistic Regression**, trained using TF-IDF features extracted from customer reviews. The model uses the importance of words in the review to classify the sentiment as Positive or Negative.

**SHAP (SHapley Additive exPlanations)** can be used to explain how individual words contribute to the model's prediction. Words with higher positive SHAP values contribute more towards a **Positive** prediction, while words with negative values contribute towards a **Negative** prediction.


````python
import shap

# Get actual TF-IDF feature names
feature_names = tfidf.get_feature_names_out()

# Create SHAP masker
masker = shap.maskers.Independent(
    X_train_tfidf,
    max_samples=100
)

# Create SHAP explainer
explainer = shap.LinearExplainer(
    best_lr,
    masker
)

# Calculate SHAP values
shap_values_raw = explainer(
    X_test_tfidf[:200]
)

# Create Explanation with actual feature names
shap_values = shap.Explanation(
    values=shap_values_raw.values,
    base_values=shap_values_raw.base_values,
    data=X_test_tfidf[:200].toarray(),
    feature_names=feature_names
)

# Plot top features
shap.plots.bar(
    shap_values,
    max_display=15
)
````

The SHAP analysis identifies the TF-IDF terms that have the greatest overall influence on the Logistic Regression sentiment model. Among the displayed features**, “not”, “good”, “great”, “awesome”**, and** “best”** have the highest mean absolute SHAP values. Other influential terms include **“worst”, “amazing”, “delicious”, “love”**, and** “excellent”. **These results indicate that the model relies heavily on sentiment-related words and negation-related terms when distinguishing between the rating-derived sentiment classes.
